## 4. Carregar dados de uma API pública

A API do **Banco Mundial** disponibiliza indicadores por país e ano. Vamos baixar dados e transformar a resposta JSON em uma tabela.

Indicadores:
- `NY.GDP.PCAP.CD`: PIB per capita em dólares correntes.
- `SP.DYN.LE00.IN`: expectativa de vida ao nascer, em anos.
- `SL.UEM.TOTL.ZS`: desemprego total estimado (% da força de trabalho).
- `SP.POP.TOTL`: população total.

**Limitações:** dólares correntes não corrigem diferenças de preços entre países nem inflação ao longo do tempo. Consulte metadados antes de usar os dados em uma pesquisa formal.

```python
BASE_URL = "https://api.worldbank.org/v2/country/all/indicator/{}"

indicadores = {
    "NY.GDP.PCAP.CD": "pib_per_capita_usd",
    "SP.DYN.LE00.IN": "expectativa_vida",
    "SL.UEM.TOTL.ZS": "desemprego_pct",
    "SP.POP.TOTL": "populacao",
}

def baixar_indicador(codigo, nome_coluna, ano_inicio=2000, ano_fim=2023):
    # Baixa um indicador e retorna país, ano e valor.
    url = BASE_URL.format(codigo)
    parametros = {
        "format": "json",
        "date": f"{ano_inicio}:{ano_fim}",
        "per_page": 20000,
    }
    resposta = requests.get(url, params=parametros, timeout=40)
    resposta.raise_for_status()
    conteudo = resposta.json()

    if not isinstance(conteudo, list) or len(conteudo) < 2 or conteudo[1] is None:
        raise ValueError(f"A API não retornou dados válidos para {codigo}")

    registros = []
    for item in conteudo[1]:
        registros.append({
            "pais": item.get("country", {}).get("value"),
            "codigo_pais": item.get("countryiso3code"),
            "regiao": item.get("region", {}).get("value"),
            "ano": int(item["date"]),
            nome_coluna: item.get("value"),
        })
    return pd.DataFrame(registros)

print("Função criada. Próximo passo: baixar os indicadores.")
```

A função ilustra `def`, argumentos, requisições HTTP, respostas JSON, um laço `for` e valores ausentes (`None`).

```python
tabelas = []
for codigo, nome_coluna in indicadores.items():
    tabela = baixar_indicador(codigo, nome_coluna)
    tabelas.append(tabela)

print("Indicadores baixados:", len(tabelas))
```

### Juntar os indicadores em uma tabela

```python
chaves = ["pais", "codigo_pais", "regiao", "ano"]
df = tabelas[0]

for tabela in tabelas[1:]:
    df = df.merge(tabela, on=chaves, how="outer")

# Remover agregados regionais; manter códigos de país com três letras.
df = df[df["codigo_pais"].str.len() == 3].copy()
df = df.sort_values(["pais", "ano"]).reset_index(drop=True)

print("Dimensões (linhas, colunas):", df.shape)
display(df.head())
```

```python
display(df.isna().sum().sort_values(ascending=False))
```

## 5. Exploração: comparar países

Vamos comparar Brasil, Chile, Portugal, México e África do Sul em 2022. Alguns indicadores podem estar ausentes.

```python
ano_escolhido = 2022
paises_foco = ["Brazil", "Chile", "Portugal", "Mexico", "South Africa"]

comparacao = df[
    (df["ano"] == ano_escolhido) &
    (df["pais"].isin(paises_foco))
].copy()

colunas = ["pais", "pib_per_capita_usd", "expectativa_vida",
           "desemprego_pct", "populacao"]
display(comparacao[colunas].sort_values("pib_per_capita_usd", ascending=False).round(2))
```

**Perguntas**
- Qual país tem maior PIB per capita nesta seleção?
- Qual tem maior expectativa de vida?
- Há dados ausentes?
- PIB per capita é o mesmo que salário médio? Por quê?

### Gráfico de dispersão: PIB per capita e expectativa de vida

```python
grafico = comparacao.dropna(subset=["pib_per_capita_usd", "expectativa_vida"])

plt.figure(figsize=(8, 5))
plt.scatter(grafico["pib_per_capita_usd"], grafico["expectativa_vida"])

for _, linha in grafico.iterrows():
    plt.annotate(linha["pais"],
                 (linha["pib_per_capita_usd"], linha["expectativa_vida"]),
                 xytext=(4, 4), textcoords="offset points")

plt.xlabel("PIB per capita (US$ correntes)")
plt.ylabel("Expectativa de vida (anos)")
plt.title(f"PIB per capita e expectativa de vida — {ano_escolhido}")
plt.grid(alpha=0.25)
plt.show()
```

**Interprete**
- Há uma tendência aparente?
- Algum país destoa?
- Quantas observações estão representadas? É suficiente para generalizar?
- Que outros fatores poderiam explicar diferenças na expectativa de vida?

Associação não é prova de causalidade.

```python
brasil = df[df["pais"] == "Brazil"].sort_values("ano").copy()

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(brasil["ano"], brasil["pib_per_capita_usd"], marker="o")
ax.set_xlabel("Ano")
ax.set_ylabel("PIB per capita (US$ correntes)")
ax.set_title("Brasil: evolução do PIB per capita")
ax.grid(alpha=0.25)
plt.show()
```

## 6. Ponte para machine learning

- **Observação:** uma linha da tabela, como um país em um ano.
- **Preditores (X):** variáveis usadas para fazer uma previsão.
- **Alvo (y):** variável que queremos prever.
- **Regressão:** prevê um valor numérico, como expectativa de vida.
- **Classificação:** prevê uma categoria.
- **Treino e teste:** dados para ajustar o modelo versus dados para avaliar o desempenho.
