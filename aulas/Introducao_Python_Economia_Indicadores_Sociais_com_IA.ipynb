## 4. Carregar dados de um arquivo CSV

Um **CSV** é um arquivo de texto em que cada linha é uma observação e os valores são separados por vírgula. É um dos formatos mais comuns para compartilhar tabelas, e o `pandas` o lê com `pd.read_csv()`.

## 5. Exploração: comparar países

Vamos comparar Brasil, Chile, Portugal, México e África do Sul em 2022. Alguns indicadores podem estar ausentes.

```python
ano_escolhido = 2022
paises_foco = ["Brazil", "Chile", "Portugal", "Mexico", "South Africa"]

comparacao = df[
    (df["ano"] == ano_escolhido) &
    (df["pais"].isin(paises_foco))
].copy()

colunas = ["pais", "pib_per_capita_usd", "expectativa_vida",
           "desemprego_pct", "populacao"]
display(comparacao[colunas].sort_values("pib_per_capita_usd", ascending=False).round(2))
```

- Qual tem maior expectativa de vida?

### Gráfico de dispersão: PIB per capita e expectativa de vida

Usamos `fig, ax = plt.subplots()`: `fig` é a figura e `ax` é o gráfico dentro dela. Em tutoriais e respostas de IA você verá também `plt.plot(...)` direto: é outro jeito de fazer a mesma coisa.

```python
grafico = comparacao.dropna(subset=["pib_per_capita_usd", "expectativa_vida"])

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(grafico["pib_per_capita_usd"], grafico["expectativa_vida"])

for _, linha in grafico.iterrows():
    ax.annotate(linha["pais"],
                (linha["pib_per_capita_usd"], linha["expectativa_vida"]),
                xytext=(4, 4), textcoords="offset points")

ax.set_xlabel("PIB per capita (US$ correntes)")
ax.set_ylabel("Expectativa de vida (anos)")
ax.set_title(f"PIB per capita e expectativa de vida — {ano_escolhido}")
ax.grid(alpha=0.25)
plt.show()
```

**Interprete**
- Há uma tendência aparente?
- Algum país destoa?
- Quantas observações estão representadas? É suficiente para generalizar?
- Que outros fatores poderiam explicar diferenças na expectativa de vida?

Associação não é prova de causalidade.

## Extra (opcional): baixar os dados de uma API

Este bloco mostra como o arquivo `dados/indicadores_banco_mundial.csv` foi criado. Não é necessário para o restante da oficina.

Uma **API** permite pedir dados a um servidor por código. A [API do Banco Mundial](https://datahelpdesk.worldbank.org/knowledgebase/articles/889392-about-the-indicators-api-documentation) devolve os indicadores em **JSON**, um formato de texto com listas e dicionários, como os da seção 2. Usamos o pacote `requests` ([documentação](https://requests.readthedocs.io/)) para fazer a requisição.

Os resultados podem diferir um pouco do arquivo da oficina, porque o Banco Mundial atualiza os dados. Por isso, este bloco salva o resultado em outro arquivo.

```python
import requests

BASE_URL = "https://api.worldbank.org/v2"

resposta = requests.get(
    f"{BASE_URL}/country/BRA/indicator/SP.DYN.LE00.IN",
    params={"format": "json", "date": "2020:2022"},
    timeout=40,
)
resposta.raise_for_status()
conteudo = resposta.json()

print(conteudo[0])     # informações da página
print(conteudo[1][0])  # primeiro registro
```

A resposta é uma lista com dois itens: informações sobre a página e os registros. Cada registro traz país, ano (`date`) e valor (`value`), que pode ser `None` quando não há dado. Ausência de dado não significa zero.

A função abaixo repete essa requisição para todos os países e transforma os registros em uma tabela.

```python
indicadores = {
    "NY.GDP.PCAP.CD": "pib_per_capita_usd",
    "SP.DYN.LE00.IN": "expectativa_vida",
    "SL.UEM.TOTL.ZS": "desemprego_pct",
    "SP.POP.TOTL": "populacao",
}

def baixar_indicador(codigo, nome_coluna, ano_inicio=2000, ano_fim=2023):
    # Baixa um indicador e retorna país, ano e valor.
    url = f"{BASE_URL}/country/all/indicator/{codigo}"
    parametros = {
        "format": "json",
        "date": f"{ano_inicio}:{ano_fim}",
        "per_page": 20000,
    }
    resposta = requests.get(url, params=parametros, timeout=40)
    resposta.raise_for_status()
    conteudo = resposta.json()

    if not isinstance(conteudo, list) or len(conteudo) < 2 or conteudo[1] is None:
        raise ValueError(f"A API não retornou dados válidos para {codigo}")

    registros = []
    for item in conteudo[1]:
        registros.append({
            "codigo_pais": item.get("countryiso3code"),
            "ano": int(item["date"]),
            nome_coluna: item.get("value"),
        })
    return pd.DataFrame(registros)

tabelas = []
for codigo, nome_coluna in indicadores.items():
    tabelas.append(baixar_indicador(codigo, nome_coluna))

print("Indicadores baixados:", len(tabelas))
```

### Países e agregados

A API também devolve **agregados** (Mundo, Renda alta, União Europeia...), que têm códigos de três letras como os países. Para separá-los, usamos a lista de países da API, em que os agregados têm região `Aggregates`.

```python
resposta = requests.get(f"{BASE_URL}/country",
                        params={"format": "json", "per_page": 400}, timeout=40)
resposta.raise_for_status()

paises_api = pd.DataFrame([
    {"codigo_pais": c["id"], "pais": c["name"], "regiao": c["region"]["value"].strip()}
    for c in resposta.json()[1]
])
paises_api = paises_api[paises_api["regiao"] != "Aggregates"]

dados = tabelas[0]
for tabela in tabelas[1:]:
    dados = dados.merge(tabela, on=["codigo_pais", "ano"], how="outer")

df_api = paises_api.merge(dados, on="codigo_pais", how="left")
df_api = df_api.sort_values(["pais", "ano"]).reset_index(drop=True)

print("Dimensões (linhas, colunas):", df_api.shape)
display(df_api.head())
```

```python
import os

os.makedirs("dados", exist_ok=True)
df_api.to_csv("dados/indicadores_banco_mundial_api.csv", index=False)
```
